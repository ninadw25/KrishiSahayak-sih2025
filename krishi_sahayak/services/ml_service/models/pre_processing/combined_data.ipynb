```python

# ==============================================================================
# Step 7: Handle remaining unmatched records
# ==============================================================================
print("\nHandling unmatched records...")

# For records that didn't match, try to find the closest match within the same state
def find_closest_soil_match(state, district, soil_df):
    """Find the closest matching district within the same state"""
    state_soil = soil_df[soil_df['State'] == state]
    if len(state_soil) == 0:
        return None
    
    # Simple similarity check - find districts with common words
    district_words = set(district.split('_'))
    best_match = None
    best_score = 0
    
    for _, row in state_soil.iterrows():
        soil_district = row['District']
        soil_words = set(soil_district.split('_'))
        
        # Calculate word overlap
        common_words = district_words & soil_words
        score = len(common_words) / max(len(district_words), len(soil_words))
        
        if score > best_score and score > 0.3:  # At least 30% similarity
            best_score = score
            best_match = row
    
    return best_match

# Apply fuzzy matching for unmatched records
if available_soil_cols and master_df[available_soil_cols[0]].isnull().sum() > 0:
    print("Applying fuzzy matching for unmatched records...")
    
    unmatched_indices = master_df[master_df[available_soil_cols[0]].isnull()].index
    matches_found = 0
    
    for idx in unmatched_indices[:100]:  # Limit to first 100 for performance
        state = master_df.loc[idx, 'State']
        district = master_df.loc[idx, 'District']
        
        match = find_closest_soil_match(state, district, soil_df)
        if match is not None:
            # Copy soil data from the matched record
            for col in soil_df.columns:
                if col not in ['State', 'District']:
                    master_df.loc[idx, col] = match[col]
            matches_found += 1
    
    print(f"Fuzzy matching found {matches_found} additional matches")
```
